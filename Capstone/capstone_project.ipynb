## Evaluation Rubric

| Criterion | 1 — Minimal | 3 — Good | 5 — Excellent |
|---|---|---|---|
| **Working system** | Errors during demo | Works on happy path | Handles edge cases gracefully |
| **Domain relevance** | Course example data | Meaningful public docs | Clearly useful, real-world domain |
| **Grounding / contrast** | Hallucinating freely | Mostly grounded / subtle difference | Faithful + cites sources / striking before/after |
| **Code quality** | One messy notebook | Organised cells, config cell | Clean, reusable functions |
| **Reflection** | 'It works' | Names one trade-off | Names failure mode + what you'd fix |

## Presentation Format (5 minutes)

1. **Problem (30 sec):** *'I built a [domain] assistant for [who].'*
2. **Architecture (60 sec):** Walk your pipeline. Track A: data → chunks → ChromaDB → retrieval → GPT-4o-mini → Gradio. Track B: dataset → NF4 base → LoRA adapters → SFTTrainer → comparison.
3. **Live demo (2 min):** 2 things that work well + 1 that doesn't.
4. **Reflection (90 sec):** What surprised you? What would you change?

Have your Gradio app **open and warmed up** before you present.
